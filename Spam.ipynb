```python
import pandas as pd
import numpy as np
```

```python
xcf = pd.read_csv('/content/emails.csv.zip')
xf = xcf.copy()
```

```python
from sklearn.model_selection import train_test_split
x = xf[['text']]
y  = xf['spam']
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vc = CountVectorizer()
x_train_vc = vc.fit_transform(x_train['text'])
x_test_vc = vc.transform(x_test['text'])
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
from sklearn.model_selection import cross_val_score
model101 = XGBClassifier(
    n_estimators=100,
    max_depth=4,
    learning_rate=0.09)
model101.fit(x_train_vc,y_train)
y10 = model101.predict(x_test_vc)
print(accuracy_score(y_test,y10))
print(confusion_matrix(y_test,y10))
score11 = cross_val_score(model101,x_train_vc,y_train,cv = 5,scoring = 'accuracy')
print(score11)
print(score11.mean())
```

```python
newemail = ['hi you won 5000']
sy = vc.transform(newemail)
pred = model101.predict(sy)
print('spam' if pred[0] == 1 else 'not spam')
```

```python
model101.save_model("spam_model.json")
```
